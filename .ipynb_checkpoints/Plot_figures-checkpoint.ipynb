```python
# Figure 2D-I
condition_ls = ['no_low', 'water_low', 'sucrose_low',
                'no_high', 'water_high', 'sucrose_high']

decision_labels = ["no reaction", "flight", "freeze", "assess+flight"]
plot_order = ["no reaction", "freeze", "assess+flight", "flight"]
decision_colors = {
    "no reaction": "mediumseagreen",
    "freeze": "royalblue",
    "assess+flight": "orange",
    "flight": "crimson"
}

summary_df = []

for cond in condition_ls:
    df_cond = df_rr[df_rr["condition"] == cond]
    for stim in sorted(df_cond["num_stim"].unique()):
        df_stim = df_cond[df_cond["num_stim"] == stim]
        n = df_stim.shape[0]
        if n == 0:
            continue
        counts = df_stim["decision_corrected"].value_counts(normalize=True) * 100
        row = {"condition": cond, "num_stim": stim}
        for i, label in enumerate(decision_labels):
            row[label] = counts.get(i, 0)  
        summary_df.append(row)

summary_df = pd.DataFrame(summary_df)
summary_df = summary_df.sort_values(["condition", "num_stim"])

fig, axes = plt.subplots(
    nrows=len(condition_ls), ncols=1,
    figsize=(10, 10), sharex=True, sharey=True
)
fig.suptitle("Decision composition across num_stim", fontsize=14, fontweight="bold")

for idx, cond in enumerate(condition_ls):
    ax = axes[idx]
    df_cond = summary_df[summary_df["condition"] == cond]
    x = df_cond["num_stim"]
    
    bottom = np.zeros_like(x, dtype=float)
    for label in plot_order:
        y = df_cond[label]
        ax.bar(
            x, y,
            bottom=bottom,
            color=decision_colors[label],
            label=label if idx == 0 else ""
        )
        bottom += y.values
    
    ax.set_ylabel(cond, rotation=0, labelpad=50, fontsize=12, va="center")
    ax.set_ylim(0, 100)

axes[-1].set_xlabel("num_stim", fontsize=12)
axes[-1].set_xticks(sorted(df_rr["num_stim"].unique()))

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, title="Decision", bbox_to_anchor=(1.02, 0.5), loc="center left")

plt.tight_layout(rect=[0, 0, 0.95, 0.96])
plt.show()

df_mean = (
    df_rr
    .groupby(["condition", "num_stim"], as_index=False)
    .mean(numeric_only=True)
)
df_sem = (
    df_rr
    .groupby(["condition", "num_stim"], as_index=False)
    .sem(numeric_only=True)
)
df_flee = df_rr[(df_rr["decision_corrected"]==1) | (df_rr["decision_corrected"]==3)]
df_flee_mean = (
    df_flee
    .groupby(["condition", "num_stim"], as_index=False)
    .mean(numeric_only=True)
)
df_flee_sem = (
    df_flee
    .groupby(["condition", "num_stim"], as_index=False)
    .sem(numeric_only=True)
)
condition_ls = ['no_low', 'water_low', 'sucrose_low',
                'no_high', 'water_high', 'sucrose_high']
feature_dict = {"duration_reward": "duration in reward zone",
                "max_instance_speed2safezone": "peak escape speed",
                "latency_to_first_escape": "latency to flee",
                "distance_under_threat":"escape distance"}
water_pal = {'no_high':"black",'water_high':"steelblue",'sucrose_high':"seagreen",
             'no_low':"darkgray",'water_low':"lightsteelblue",'sucrose_low':"darkseagreen"}   
colors = [water_pal.get(key, None) for key in condition_ls] 

for i_feature in ["duration_reward","max_instance_speed2safezone", "distance_under_threat"]:

    plt.figure(figsize=(8, 4))
    for color, condition in zip(colors, condition_ls):
        df_condition = df_mean[df_mean["condition"] == condition]
        df_condition_sem = df_sem[df_sem["condition"] == condition]
        plt.plot(
            df_condition["num_stim"],
            df_condition[i_feature],
            "-o",
            label=condition,
            color=color,
            alpha=1
        )
        plt.fill_between(
            df_condition_sem["num_stim"], 
            df_condition[i_feature]-df_condition_sem[i_feature], 
            df_condition[i_feature]+df_condition_sem[i_feature],
            color=color, alpha=0.2)

    plt.ylabel(feature_dict[i_feature])
    plt.xlabel("Number of trial")
    plt.title(feature_dict[i_feature])
    plt.xlim([0.5, 10.5])
    plt.legend(title="Condition", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.show()
    
for i_feature in ["latency_to_first_escape"]:
    plt.figure(figsize=(8, 4))
    for color, condition in zip(colors, condition_ls):
        df_condition = df_flee_mean[df_flee_mean["condition"] == condition]
        df_condition_sem = df_flee_sem[df_flee_sem["condition"] == condition]
        plt.plot(
            df_condition["num_stim"],
            df_condition[i_feature],
            "-o",
            label=condition,
            color=color,
            alpha=1
        )
        plt.fill_between(
            df_condition_sem["num_stim"], 
            df_condition[i_feature]-df_condition_sem[i_feature], 
            df_condition[i_feature]+df_condition_sem[i_feature],
            color=color, alpha=0.2)

    plt.ylabel(feature_dict[i_feature])
    plt.xlabel("Number of trial")
    plt.title(feature_dict[i_feature])
    plt.xlim([0.5, 10.5])
    plt.legend(title="Condition", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.show()

    df_change = df_rr.groupby("mice_id").agg({
    "condition": "first",
    "risk": "first",
    "reward": "first",
    "change_trial": "first",
})
df_change = df_change[df_change.index!="idm41_CE93"]

condition_ls = ['no_low', 'water_low', 'sucrose_low',
                'no_high', 'water_high', 'sucrose_high']
water_pal = {'no_high':"darkgray",'water_high':"steelblue",'sucrose_high':"seagreen",
             'no_low':"darkgray",'water_low':"steelblue",'sucrose_low':"seagreen"}

feature_ls = ["change_trial"]
for i_feature in feature_ls:
    fig, ax = plt.subplots(figsize=(2, 6))
    data_to_plot = [df_change[df_change["condition"] == cond][i_feature].dropna()
                    for cond in condition_ls]

    box = ax.boxplot(
        data_to_plot,
        positions=np.arange(len(condition_ls)),
        vert=False,
        patch_artist=True,
        showfliers=False,
        widths=0.6
    )

    for patch, cond in zip(box['boxes'], condition_ls):
        patch.set_facecolor("none")
        patch.set_edgecolor(water_pal[cond])
        patch.set_linewidth(0.5)
        patch.set_zorder(2)

    for median in box['medians']:
        median.set_color('red')
        median.set_linewidth(1)
        median.set_zorder(3)

    for whisker, cond in zip(box['whiskers'],np.repeat(condition_ls, 2)):
        whisker.set_color(water_pal[cond])
        whisker.set_linewidth(0.5)

    for cap, cond in zip(box['caps'],np.repeat(condition_ls, 2)):
        cap.set_color(water_pal[cond])
        cap.set_linewidth(1)
        
    for idx, cond in enumerate(condition_ls):
        vals = df_change[df_change["condition"] == cond][i_feature].dropna()
        y_jitter = np.random.normal(loc=idx, scale=0.15, size=len(vals))  # jitter on y-axis
        ax.scatter(
            vals, y_jitter,
            s=15, alpha=1, color="black",linewidths=0,zorder=10
        )

    ax.invert_yaxis()  

    ax.tick_params(width=0.5)
    for spine in ["top","right"]:
        ax.spines[spine].set_visible(False)
    for spine in ["bottom","left"]:
        ax.spines[spine].set_linewidth(0.5)

    ax.set_axisbelow(True)
    ax.set_yticks(np.arange(len(condition_ls)))
    ax.set_yticklabels(condition_ls)
    plt.xlim([2.5,10])
    plt.xticks([3,6,9])
    plt.xlabel(i_feature, fontsize=12)
    plt.show()
```

```python
# Figure 4I
mouse_ranked_ls = [["idm25_8033", "idm25_81A3"],
                  ["idm26_1_87F3","idm26_1_A5F3"],
                  ["idm26_2_8663","idm26_2_A7C3"],
                  ["idm26_5_0873","idm26_5_85C3"],
                  ["idm26_6_8153","idm26_6_84A3"],
                   ["idm33_1_AAF3", "idm33_1_A3A3"],
                  ["idm33_3_93F3", "idm33_3_90A3"],
                  ["idm33_4_9A53", "idm33_4_95A3"],
                  ["idm33_5_9B33", "idm33_5_0933"]]

mouse_dom_ls = []
mouse_sub_ls = []
for i in range(len(mouse_ranked_ls)):
    mouse_dom_ls.append(mouse_ranked_ls[i][0])
    mouse_sub_ls.append(mouse_ranked_ls[i][1])

condition_ls = ['dom','sub']
table_count = pd.DataFrame(index=[0,1],columns=["no reaction","freeze","assess+flight","flight"])
for i in range(len(condition_ls)):
    i_con = condition_ls[i]
    df_n = df_social[df_social["social_rank"]==i_con]
    num_trial_n = df_n.shape[0]
    table_count.loc[i,"no reaction"] = df_n[df_n["decision_corrected"]==0].shape[0]
    table_count.loc[i,"flight"] = df_n[df_n["decision_corrected"]==1].shape[0]
    table_count.loc[i,"freeze"] = df_n[df_n["decision_corrected"]==2].shape[0]
    table_count.loc[i,"assess+flight"] = df_n[df_n["decision_corrected"]==3].shape[0]

label_order = ["flight","assess+flight","freeze","no reaction"]
color_order = ["crimson","orange","royalblue","mediumseagreen"]
table_t = table_count.T
table_t = table_t.reindex(label_order,fill_value=0)
for i in range(2):
    plt.figure(figsize=(3,3))
    plt.pie(table_t[i], colors=color_order)
    plt.tight_layout(pad=3)
    plt.show()

df_pivot = df_social.pivot(index="pair_trial", columns="social_rank", values="decision_corrected")

pairs = df_pivot.values
contingency = np.zeros((4, 4))

for dom_decision, sub_decision in pairs:
    contingency[int(dom_decision), int(sub_decision)] += 1

print("Contingency table:")
print(contingency)

# Stuart-Maxwell test
X2 = 0
df_test = 0
for i in range(4):
    for j in range(i+1, 4):
        b_ij = contingency[i, j]
        b_ji = contingency[j, i]
        denom = b_ij + b_ji
        if denom > 0:
            X2 += (b_ij - b_ji)**2 / denom
            df_test += 1

pval = 1 - chi2.cdf(X2, df_test)

print(f"Stuart-Maxwell X² = {X2:.4f}, df = {df_test}, p = {pval}")
```

```python
# Figure S9H-K
alpha = 1.78
x0 = 0
threshold = 0.71
decision_time_limit = 2.0

r_vals = np.linspace(0, 1, 501)
beta_values = [1, 1.2, 1.5, 2] 

# Initialize arrays for results
decision_grid = np.zeros((len(beta_values), len(r_vals)))
latency_grid = np.full((len(beta_values), len(r_vals)), np.nan) 

# Calculate decision and latency for each beta and r
print("Calculating decision and latency for each (β, r)...")
for i, beta in enumerate(beta_values):
    for j, r in enumerate(r_vals):
        t_eval = np.linspace(0, decision_time_limit, 201)
        x_eval = x_analytical(t_eval, alpha, beta, r, x0)
        
        # Decision: 1 if any x > 0.71 before decision_time_limit
        decision_grid[i, j] = 1 if np.any(x_eval > threshold) else 0

        latency = find_latency(alpha, beta, r, x0, threshold, t_max=10)
        if latency is not None:
            latency_grid[i, j] = latency

print("Calculation complete!")

# Create grid for heatmaps with r in [0,1] and beta in [1,2]
beta_heat = np.linspace(1, 2, 101) 
r_heat = np.linspace(0, 1, 101)
beta_grid, r_grid = np.meshgrid(beta_heat, r_heat)

# Initialize arrays for heatmaps
decision_heat = np.zeros(beta_grid.shape)
latency_heat = np.full(beta_grid.shape, np.nan)

# Calculate decision and latency on grid
print("Calculating heatmap data...")
for i in range(beta_grid.shape[0]):
    for j in range(beta_grid.shape[1]):
        beta_val = beta_grid[i, j]
        r_val = r_grid[i, j]
        
        # Check decision
        t_eval = np.linspace(0, decision_time_limit, 201)
        x_eval = x_analytical(t_eval, alpha, beta_val, r_val, x0)
        decision = 1 if np.any(x_eval > threshold) else 0
        decision_heat[i, j] = decision  # 0 or 1
        
        # Calculate latency
        latency = find_latency(alpha, beta_val, r_val, x0, threshold, t_max=10)
        if latency is not None:
            latency_heat[i, j] = latency

print("Heatmap calculation complete!")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Heatmap of decision (0 or 1)
binary_cmap = ListedColormap(['mediumseagreen', 'crimson'])

heatmap1 = axes[0, 0].imshow(decision_heat, 
                          extent=[beta_grid.min(), beta_grid.max(), 
                                  r_grid.min(), r_grid.max()],
                          origin='lower',
                          aspect='auto',
                          cmap=binary_cmap,
                          vmin=0, vmax=1)

axes[0, 0].set_xlabel('β', fontsize=12)
axes[0, 0].set_ylabel('r', fontsize=12)
axes[0, 0].set_title(f'Decision Heatmap\n(1 if x > {threshold} before t={decision_time_limit})', fontsize=12)

cbar1 = plt.colorbar(heatmap1, ax=axes[0, 0], ticks=[0.25, 0.75])
cbar1.set_ticklabels(['0 (No)', '1 (Yes)'])
cbar1.set_label('Decision', fontsize=12)

# Heatmap of latency
latency_mask = ~np.isnan(latency_heat)
if np.any(latency_mask):
    latency_plot = np.copy(latency_heat)
    latency_plot[np.isnan(latency_plot)] = 10
    
    heatmap2 = axes[0, 1].imshow(latency_plot, 
                              extent=[beta_grid.min(), beta_grid.max(), 
                                      r_grid.min(), r_grid.max()],
                              origin='lower',
                              aspect='auto',
                              cmap='viridis_r',
                              vmin=1, vmax=3)
    
    axes[0, 1].set_xlabel('β', fontsize=12)
    axes[0, 1].set_ylabel('r', fontsize=12)
    axes[0, 1].set_title(f'Latency Heatmap\n(time when x > {threshold})', fontsize=12)
    cbar2 = plt.colorbar(heatmap2, ax=axes[0, 1], label='Latency')
    
    contour_limit = axes[0, 1].contour(beta_grid, r_grid, latency_plot, 
                                    levels=[decision_time_limit], colors='red', linewidths=1)
    
    nan_regions = np.isnan(latency_heat)
    if np.any(nan_regions):
        nan_indices = np.where(nan_regions)
        if len(nan_indices[0]) > 0:
            rep_idx = (nan_indices[0][0], nan_indices[1][0])
            rep_beta = beta_grid[rep_idx]
            rep_r = r_grid[rep_idx]
            axes[0, 1].text(rep_beta, rep_r, 'Never reaches\nthreshold', 
                         ha='center', va='center', fontsize=9,
                         bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
else:
    axes[0, 1].text(0.5, 0.5, 'No latency data\n(never reaches threshold)', 
                 ha='center', va='center', transform=axes[0, 1].transAxes,
                 fontsize=12, bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
    axes[0, 1].set_xlabel('β', fontsize=12)
    axes[0, 1].set_ylabel('r', fontsize=12)
    axes[0, 1].set_title(f'Latency Heatmap\n(time when x > {threshold})', fontsize=12)

# Latency vs r for different beta
ax3 = axes[1, 0]
for i, beta in enumerate(beta_values):
    valid_indices = ~np.isnan(latency_grid[i])
    if np.any(valid_indices):
        ax3.plot(r_vals[valid_indices], latency_grid[i][valid_indices], 
                linewidth=1, label=f'β={beta}', marker='.', markersize=3)

ax3.set_xlabel('r', fontsize=12)
ax3.set_ylabel('Latency (time when x > 0.71)', fontsize=12)
ax3.set_title('Latency vs r', fontsize=12)
ax3.legend(fontsize=10)
ax3.set_ylim(bottom=0)

# Latency vs beta for different r
beta_range = np.linspace(1, 2, 301)  
r_latency = [0, 0.3, 0.6, 0.9] 

ax4 = axes[1, 1]
for idx, r in enumerate(r_latency):
    latencies = []
    for beta in beta_range:
        latency = find_latency(alpha,beta, r, x0, threshold, t_max=10)
        if latency is not None:
            latencies.append(latency)
        else:
            latencies.append(np.nan)
    
    latencies = np.array(latencies)
    valid = ~np.isnan(latencies)
    
    if np.any(valid):
        ax4.plot(beta_range[valid], latencies[valid], 
                linewidth=1, label=f'r={r}', marker='.', markersize=3)

ax4.set_xlabel('β', fontsize=12)
ax4.set_ylabel('Latency (time when x > 0.71)', fontsize=12)
ax4.set_title('Latency vs β for different r values', fontsize=12)
ax4.legend(fontsize=10)

plt.tight_layout()
plt.show()
```
